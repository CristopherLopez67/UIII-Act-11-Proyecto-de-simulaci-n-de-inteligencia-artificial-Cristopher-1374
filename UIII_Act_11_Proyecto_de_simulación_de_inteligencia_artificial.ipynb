#### Taxista: Número de unidad, zona de operación, plataforma digital, marca del auto, calificación de usuario y turno.

```python
print("==============================================")
print("   SISTEMA IA DE GESTION DE TAXISTAS")
print("==============================================")

# Lista vacía de diccionarios
taxistas = []

# Función para agregar taxistas
def agregar_taxista():
    print("\n--- REGISTRO DE TAXISTA ---")

    numero_unidad = input("Número de unidad: ")
    zona_operacion = input("Zona de operación: ")
    plataforma = input("Plataforma digital: ")
    marca_auto = input("Marca del auto: ")
    calificacion = input("Calificación de usuario: ")
    turno = input("Turno: ")

    taxista = {
        "Número de unidad": numero_unidad,
        "Zona de operación": zona_operacion,
        "Plataforma digital": plataforma,
        "Marca del auto": marca_auto,
        "Calificación": calificacion,
        "Turno": turno
    }

    taxistas.append(taxista)

    print("\n✅ Taxista registrado correctamente.")


# Función para mostrar taxistas
def mostrar_taxistas():

    if len(taxistas) == 0:
        print("\n⚠ No hay taxistas registrados.")

    else:
        print("\n================ LISTA DE TAXISTAS ================")

        print("{:<18} {:<20} {:<20} {:<18} {:<15} {:<10}".format(
            "Unidad",
            "Zona",
            "Plataforma",
            "Auto",
            "Calificación",
            "Turno"
        ))

        print("-" * 105)

        for t in taxistas:
            print("{:<18} {:<20} {:<20} {:<18} {:<15} {:<10}".format(
                t["Número de unidad"],
                t["Zona de operación"],
                t["Plataforma digital"],
                t["Marca del auto"],
                t["Calificación"],
                t["Turno"]
            ))


# Función para buscar taxista
def buscar_taxista():

    if len(taxistas) == 0:
        print("\n⚠ No hay datos registrados.")

    else:
        buscar = input("\nIngresa el número de unidad: ")

        encontrado = False

        for t in taxistas:

            if t["Número de unidad"] == buscar:

                print("\n===== TAXISTA ENCONTRADO =====")
                print("Número de unidad:", t["Número de unidad"])
                print("Zona de operación:", t["Zona de operación"])
                print("Plataforma digital:", t["Plataforma digital"])
                print("Marca del auto:", t["Marca del auto"])
                print("Calificación:", t["Calificación"])
                print("Turno:", t["Turno"])

                encontrado = True

        if encontrado == False:
            print("\n❌ Taxista no encontrado.")


# Función para análisis IA simple
def analisis_ia():

    if len(taxistas) == 0:
        print("\n⚠ No hay registros para analizar.")

    else:

        print("\n=========== ANALISIS IA ===========")

        for t in taxistas:

            if float(t["Calificación"]) >= 4.8:
                print(
                    "⭐ El taxista de la unidad",
                    t["Número de unidad"],
                    "tiene excelente servicio."
                )

            elif float(t["Calificación"]) >= 4.0:
                print(
                    "👍 El taxista de la unidad",
                    t["Número de unidad"],
                    "tiene buen rendimiento."
                )

            else:
                print(
                    "⚠ El taxista de la unidad",
                    t["Número de unidad"],
                    "debe mejorar su servicio."
                )


# Menú principal
while True:

    print("\n============== MENU ==============")
    print("1. Registrar taxista")
    print("2. Mostrar taxistas")
    print("3. Buscar taxista")
    print("4. Analisis IA")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        agregar_taxista()

    elif opcion == "2":
        mostrar_taxistas()

    elif opcion == "3":
        buscar_taxista()

    elif opcion == "4":
        analisis_ia()

    elif opcion == "5":
        print("\n👋 Saliendo del sistema...")
        break

    else:
        print("\n❌ Opción inválida.")
```
